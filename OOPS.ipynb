```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(self.brand, "is driving")


car1 = Car("Toyota", "Red")
car2 = Car("BMW", "Black")

car1.drive()
car2.drive()
```

```python
# Car is a class.
# car1 and car2 are objects (instances of Car).
# brand and color are attributes.
# drive() is a method.
# __init__() is the constructor.
# self refers to the current object.

class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(self.brand, "is driving")


car1 = Car("Toyota", "Red")
car2 = Car("BMW", "Black")

car1.drive()
car2.drive()
```
