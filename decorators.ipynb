```python
# Decorator function, Wrapper running"
def my_decorator(func):
    print("Decorator running")

    def wrapper():
        print("Wrapper running")
        func()

    return wrapper
@my_decorator
def hello():
    print("Hello")
```

```python
# A closure is a function that:

# Is defined inside another function.
# Uses a variable from the outer function.
# Remembers that variable even after the outer function has finished.

# The key idea is:

# A closure allows an inner function to remember values from its enclosing function.

def outer():
    x = 10

    def inner():
        print(x)

    inner()

outer()

# belongs to outer().
# And inner() can access x.
# That's because inner() is inside outer().
```

```python
def decorator(func):

    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result

    return wrapper
```

```python
class MyNumbers:

    def __init__(self):
        self.number = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.number <= 3:
            value = self.number
            self.number += 1
            return value

        raise StopIteration
numbers = MyNumbers()
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

```python

# “Lambda is an anonymous function used for small, one-line operations. 
# map() is used to apply a function to every element of an iterable. 
# filter() is used to select elements based on a condition. 
# reduce() is used to combine all elements into a single result.”from functools import reduce

numbers = [1, 2, 3, 4, 5]

# lambda
square = lambda x: x * x

# map
result = map(lambda x: x * 2, numbers)

# filter
even = filter(lambda x: x % 2 == 0, numbers)

# reduce
total = reduce(lambda a, b: a + b, numbers)
```

```python
salaries = [30000, 40000, 50000, 60000]

# Increase each salary by 10%
new_salaries = list(map(lambda x: x * 1.10, salaries))

# Salaries greater than 45000
high_salaries = list(filter(lambda x: x > 45000, salaries))

# Total salary
from functools import reduce
total = reduce(lambda a, b: a + b, salaries)
print("New Salaries:", new_salaries)
print("High Salaries:", high_salaries)
print("Total Salary:", total)
```
