```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "Chenna", "Kesava", "Reddy"],
    "Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 55000],
    "state": ['Bangalore','HYD','chennai','HYD','Bangalore','HYD']
}
df = pd.DataFrame(data)

print(df[["Name", "Salary"]])
```

```python
df.loc[:, ["Name", "Salary"]]
```

```python
result = df[df["Salary"] > 60000]

print(result)
```

```python
#Below is the code to filter the rows where the salary is between 50000 and 70000.
result = df[df["Salary"].between(50000, 70000)]

print(result)
```

```python
#Bonus calculation.
df["Bonus"] = df["Salary"] * 0.10
print(df)
```

```python
#update the salary of all employees by adding 5000 to their current salary.
df["Salary"] = df["Salary"] + 5000
print(df)
```

```python
#Drop the Bonus column from the DataFrame.
df.drop("Bonus", axis=1, inplace=True)
print(df)
```

```python
#Drop the first row from the DataFrame.
df1=df.drop(0, axis=0)
print(df1)
```

```python
#Sort the DataFrame by the "Salary" column in ascending order.
df.sort_values("Salary")
ss=df.sort_values("Salary", ascending=False)
print(ss)
```

```python
#Increase the salary of all employees by 10% using the apply() function.
s=df["Name"].str.upper()
print(s)
```

```python
#Increase the salary of all employees by 10% using the apply() function.
chenna=df["Salary"] = df["Salary"].apply(lambda x: x * 1.10)
print(chenna)
```

```python
#Increase the salary of all employees by 10% using a custom function.
def increase_salary(x):
    return x * 1.10

emp=df["Salary"] = df["Salary"].apply(increase_salary)
print(emp)
```

```python
#Group the DataFrame by the "Department" column and calculate the mean salary for each department.
groupby=df.groupby("Department")["Salary"].mean()
groupbt1=df.groupby("Department")["Salary"].max()
print(groupby)
print(groupbt1)
```

```python
#Group the DataFrame by the "Department" column and calculate the mean, min, max, and count of salaries for each department using a pivot table.
res = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc=["mean", "min", "max",'count']
)

print(res)
```

```python
# or Group the DataFrame by the "Department" column and calculate the mean, min, max, and count of salaries for each department using the groupby() method.
multi=df.groupby("Department")["Salary"].agg(
    ["mean", "min", "max"]
    
)
print(multi)
```
